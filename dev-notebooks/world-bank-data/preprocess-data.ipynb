```python
import pandas as pd
```

```python
f_gdp = 'raw/API_NY.GDP.PCAP.CD_DS2_en_csv_v2_358064/API_NY.GDP.PCAP.CD_DS2_en_csv_v2_358064.csv'
f_pop = 'raw/API_SP.POP.TOTL_DS2_en_csv_v2_346039/API_SP.POP.TOTL_DS2_en_csv_v2_346039.csv'
```

```python
gdp = pd.read_csv(f_gdp, quotechar='"', skiprows=4)
pop = pd.read_csv(f_pop, quotechar='"', skiprows=4)
```

```python
gdp = gdp[['Country Name', 'Country Code', '2024']]
```

```python
pop = pop[['Country Code', '2024']]
```

```python
gdp.rename(columns={'2024': 'GDP-2024'}, inplace=True)
pop.rename(columns={'2024': 'Pop-2024'}, inplace=True)
```

```python
df = gdp.merge(pop, on='Country Code', how='inner')
```

```python
df = df.dropna(subset=['GDP-2024', 'Pop-2024'])
```
