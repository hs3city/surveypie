# Gini Index

Gini Index is used to measure an inequality in a population. This index uses continuous and positive data like income (*), and it won't work well with ordinal data. This index is often used in economic studies because of it's simplicity. When Gini Index is close to zero then inequality is low, and Gini Index close to one means that inequality is high. There are no parameters to set in Gini Coefficient.

**Note 1**: this index shouldn't be used with ordinal variables!

## Formulas

The basic formula for Gini Index, proposed in (1) is defined as:

$$G = \frac{\sum_{i=1}^{n}\sum_{j=1}^{n}|x_{i}-x_{j}|}{2n^2\mu}$$

where:

- $x_{i}$ and $x_{j}$ are the individual values from provided dataset
- $n$ is a total number of values
- $\mu$ is the average (mean) of the value in input data

The package uses slightly changed formula that makes computation more efficient (2):

$$G = \frac{2\sum_{i=1}^{n}i*x_{i}}{n\sum_{i=1}^{n}x_{i}}-\frac{n+1}{n}$$

where:

- data array $x$ must be sorted in ascending order
